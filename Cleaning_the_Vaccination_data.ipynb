# **Project Summary -**


**Vaccination Data Analysis and Visualization** is a public health analytics project focused on analyzing global vaccination coverage, disease incidence, reported cases, vaccine introduction, and vaccination schedules. The project aims to identify vaccination trends, regional disparities, and relationships between vaccination coverage and disease reduction.

The project involves **data cleaning, exploratory data analysis (EDA), SQL database management, and Power BI visualization**. Raw vaccination and disease-related data is cleaned by handling missing values, standardizing units, and maintaining consistent date formats. The cleaned data is then organized into normalized relational SQL tables with appropriate primary and foreign-key relationships.

Power BI is connected to the SQL database to create interactive dashboards containing **KPI indicators, geographical heatmaps, trend charts, bar charts, and scatter plots**. These visualizations help analyze vaccination coverage over time, disease incidence across regions, and the relationship between vaccination rates and disease outcomes.

The analysis explores important public health questions, including vaccination coverage disparities, first-dose to subsequent-dose drop-off, booster uptake, urban-rural differences, vaccine introduction timelines, and disease reduction following vaccination campaigns. The project also supports practical use cases such as identifying low-coverage regions, improving resource allocation, estimating vaccine demand, and evaluating vaccination strategies.

The final deliverables include **Python scripts for data cleaning and extraction, SQL queries and a structured database, interactive Power BI dashboards, and project documentation**. Overall, the project demonstrates how data analytics and visualization can transform vaccination and disease data into meaningful insights for public health planning and decision-making.

```python
# ============================================================
# BUILD SQL DATABASE
# ============================================================

import sqlite3
import pandas as pd
from pathlib import Path


# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

CLEAN_DIR = Path(
    "/content/drive/MyDrive/Labmentix Projects/"
    "Vaccination_Data_Analysis/cleaned_data"
)

DB_PATH = CLEAN_DIR / "vaccination.db"

print("Clean data directory:")
print(CLEAN_DIR)

print("\nDatabase:")
print(DB_PATH)


# ------------------------------------------------------------
# 2. LOAD CLEANED CSV FILES
# ------------------------------------------------------------

tables = {
    "coverage": pd.read_csv(
        CLEAN_DIR / "coverage_clean.csv"
    ),

    "schedule": pd.read_csv(
        CLEAN_DIR / "vaccine_schedule_clean.csv"
    ),

    "introduction": pd.read_csv(
        CLEAN_DIR / "vaccine_introduction_clean.csv"
    ),

    "cases": pd.read_csv(
        CLEAN_DIR / "reported_clean.csv"
    ),

    "incidence": pd.read_csv(
        CLEAN_DIR / "incidence_clean.csv"
    )
}


# ------------------------------------------------------------
# 3. DISPLAY DATASET INFORMATION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("CLEANED DATASETS")
print("=" * 60)

for name, df in tables.items():
    print(
        f"{name:20} : "
        f"{df.shape[0]:,} rows × "
        f"{df.shape[1]} columns"
    )
```

```python
for name, df in tables.items():

    print("\n" + "=" * 60)
    print(name.upper())
    print("=" * 60)

    print(df.columns.tolist())
```

```python
def standardize_columns(df):

    df = df.copy()

    df.columns = (
        df.columns
        .astype(str)
        .str.strip()
        .str.lower()
        .str.replace(" ", "_", regex=False)
        .str.replace("-", "_", regex=False)
    )

    return df


for name in tables:

    tables[name] = standardize_columns(
        tables[name]
    )
```
